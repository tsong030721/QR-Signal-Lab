```python
import pandas as pd
from qr_signal_lab.feature.pipeline import compute_features
from qr_signal_lab.strategy.pipeline import compute_positions

# Compute features
tickers = ['CL=F', 'GC=F', 'HG=F', 'NG=F', 'ZC=F']
data = {}
for ticker in tickers:
    data[ticker] = pd.read_parquet(f"../data/clean/{ticker}.parquet")
features = compute_features(data, True)

positions = compute_positions(features, True)
```

```python
from qr_signal_lab.backtest.engine import compute_returns, compute_portfolio_weights, compute_gross_returns, compute_net_returns, compute_equity_curve
from qr_signal_lab.backtest.costs import compute_turnovers, transaction_costs
returns = features["pct_return"].iloc[1:,:]
positions_aligned = positions["momentum_return"].shift(1).iloc[1:,:]


raw_returns = compute_returns(returns, positions_aligned)
portfolio = compute_portfolio_weights(positions_aligned)

turnovers = compute_turnovers(portfolio)
costs = transaction_costs(turnovers, 2)

gross_returns = compute_gross_returns(portfolio, returns)
net_returns = compute_net_returns(gross_returns, costs)
equity = compute_equity_curve(net_returns, 100000)

equity
```
